使用者輸入：請輸入正整數$n$，其中$1\le n\le 9$，輸出$nn$乘法表。格式如Tranclass的圖片Exercise20261008.png，以$n=5$為例。

1. 只能使用for。
3. 字串格式化只能使用 f-string 對齊。

```python
# 輸入正整數 n (1 <= n <= 9)
n = int(input("請輸入正整數 n (1<=n<=9): "))

# 使用巢狀 for 迴圈逐行、逐列印出乘法表
for i in range(1, n + 1):
  row = []
  for j in range(1, n + 1):
    # 使用 f-string 將十進位轉為二進位，並指定位數補零對齊
    # :03b 表示 3位二進位，不足補0；:06b 表示 6位二進位，不足補0
    expression = f"{i:03b}×{j:03b}={i*j:06b}"
    row.append(expression)

  # 將每一行的各個算式用空格連接並輸出
  print("   ".join(row))
```

使用者輸入：請輸入正整數$n$，其中$1\le n\le 9$，輸出二進位$nn$乘法表。格式如如Tranclass的圖片Exercise20261008.png，以$n=5$為例。                                                                                                         
                                                                                                            
1. 只能使用while。                                                                                            
3. 字串格式化只能使用 f-string 對齊。                                                                       

```python
# 輸入正整數 n (1 <= n <= 9)
n = int(input("請輸入正整數 n (1<=n<=9): "))

# 使用 while 迴圈控制列數 (i)
i = 1
while i <= n:
  row = []
  j = 1
  # 使用內層 while 迴圈控制行數 (j)
  while j <= n:
    # 使用 f-string 將十進位轉為二進位，並指定位數補零對齊
    # :03b 表示 3位二進位，不足補0；:06b 表示 6位二進位，不足補0
    expression = f"{i:03b}×{j:03b}={i*j:06b}"
    row.append(expression)
    j += 1

  # 將每一行的各個算式用空格連接並輸出
  print("   ".join(row))
  i += 1
```
