## 第四部分：深入探討各種型別

請針對第三部分中的每一種型別回答：

1. 它是否為 Python 的內建型別？
2. 對應的 CPython 物件結構名稱為何？
3. 它如何儲存資料？
4. 是否有固定大小限制？
5. 若資料規模持續增加，如何擴充儲存空間？

1. int（自然數 / 整數）
是否為內建型別：是。

CPython 物件結構名稱：PyLongObject。

如何儲存資料：使用變長數位陣列（digit array）結合正負符號位來儲存。

是否有固定大小限制：無位元大小限制（支援任意精度，僅受限於電腦記憶體）。

規模增加如何擴充：當數值變大而原空間不足時，會動態重新配置記憶體以擴大底層數位陣列。

2. float（實數）
是否為內建型別：是。

CPython 物件結構名稱：PyFloatObject。

如何儲存資料：直接包裝 C 語言的 64 位元雙精度浮點數（double）。

是否有固定大小限制：有（固定佔用 8 個位元組）。

規模增加如何擴充：不適用（大小固定，無法擴充）。

3. complex（複數）
是否為內建型別：是。

CPython 物件結構名稱：PyComplexObject。

如何儲存資料：內部包含兩個 C 語言的 double，分別代表實部（real）與虛部（imag）。

是否有固定大小限制：有（固定大小）。

規模增加如何擴充：不適用（大小固定，無法擴充）。

4. set（集合）
是否為內建型別：是。

CPython 物件結構名稱：PySetObject。

如何儲存資料：基於雜湊表（Hash Table）架構，儲存指向各個鍵（Key）的指標。

是否有固定大小限制：無硬性大小限制（受限於記憶體）。

規模增加如何擴充：當負載因子過高時，會自動進行擴容（通常為容量翻倍），並重新計算所有元素的雜湊位置（Rehashing）。

5. bool（命題真假）
是否為內建型別：是。

CPython 物件結構名稱：PyBoolObject（為 int 的子類別）。

如何儲存資料：記憶體中僅存在 True 與 False 兩個全域單例（Singleton），本質上共用整數 1 與 0。

是否有固定大小限制：有（全域就只有這兩個固定物件）。

規模增加如何擴充：不適用（不需擴充）。

6. list（串列）
是否為內建型別：是。

CPython 物件結構名稱：PyListObject。

如何儲存資料：使用一段連續的記憶體空間，儲存指向各個 Python 物件的「指標陣列」。

是否有固定大小限制：無硬性大小限制（受限於記憶體）。

規模增加如何擴充：採用超額配置（Over-allocation）策略。當容量滿時，系統會自動分配一塊更大的記憶體空間（通常增加約 1/8 或依特定公式成長），以減少頻繁重新配置記憶體的開銷。

## 第五部分：Python 整數與浮點數

執行下列程式：

```python
print(10**100)
print(10**1000)
```

回答：

1. 為什麼 Python 能表示如此大的整數？
    因為 Python 內建的整數型別支援動態記憶體配置。當數字變得非常大（超過傳統 32 或 64 位元暫存器上限）時，Python 會自動增加記憶體空間來儲存更多的位數，而不是直接發生溢位（Overflow）。
2. Python 的 `int` 是否具有固定的位元數限制？
    沒有。與 C 或 Java 等語言有固定的 int32、int64 限制不同，Python 的整數沒有固定的位元數上限。
3. 什麼是 Arbitrary Precision（任意精度）？
    這是一種數值表示法。系統不會受限於固定的位元大小，而是根據數字的大小自動調整（擴充）所需的記憶體空間，能精確表示任意大小的整數，不會有精度遺失的問題。
4. CPython 使用什麼物件來實作 `int`？
    使用 PyLongObject 結構。它包含一個計數器、正負符號標記，以及一個用來儲存數值片段（digits）的動態陣列。
5. Python 整數大小真正的限制是什麼？
    真正的限制是電腦可用的實體記憶體（RAM）與虛擬記憶體空間大小。只要記憶體夠大，你想算多大的數字都可以；另外在字串轉換方面，Python 也有為了防範阻斷服務攻擊而設置的位數上限保護（可透過 sys.set_int_max_str_digits() 調整）。

回答：

1. 為什麼結果與數學上的答案不同？
    因為電腦採用二進位無法完美轉換多數十進位小數，會產生微小的捨入誤差。
2. 浮點數是否能精確表示所有小數？
    不行，受限於二進位與位元數有限，絕大多數小數只能被近似表示。
3. 什麼是浮點數的精度（precision）？
    是指數字能被記錄的有效位數與細膩程度。
4. 什麼是浮點數的範圍（range）？
    是指系統能表示的數值大小極限。
5. 精度與範圍有何不同？
    精度決定了數字能切得有多細，而範圍決定了數字可以有多巨大或多接近零。

## 第六部分：AI 模型中的數值表示

人工智慧模型需要儲存大量參數（Parameters）與權重（Weights）。

請查閱資料並回答：

1. AI 模型中的權重通常使用什麼型別儲存？
    AI 模型中的權重通常使用浮點數型別（如 FP32、FP16 或 BF16）來儲存，因為神經網路的訓練與推論需要大量的連續小數與梯度計算。
2. 什麼是 FP64？
    FP64 是 64 位元雙精度浮點數，能提供極高的精確度與極大的數值範圍，但因運算速度慢且佔用大量記憶體，通常用於高精度科學計算而非一般 AI 模型。
3. 什麼是 FP32？
    FP32 是 32 位元單精度浮點數，過去是深度學習訓練與推論的標準型別，能夠提供足夠的精確度來確保模型穩定收斂。
4. 什麼是 FP16？
    FP16 是 16 位元半精度浮點數，透過縮減位元數來節省記憶體並大幅加快硬體運算速度，但因為數值表示範圍較小，訓練時容易發生溢位或精度不足。
5. 什麼是 BF16？
    BF16 是 Google 開發的 16 位元浮點數格式，它保留了與 FP32 相同的數值範圍（指數位較多），僅縮減了精度（尾數位較少），非常適合用於大型 AI 模型訓練以防止數值溢位。
6. 為什麼大型 AI 模型不直接使用 Python 的 `int` 作為權重？
    因為 Python 的 int 是支援任意精度的動態物件，會佔用過多記憶體，而神經網路需要小數點來進行微調與梯度下降；若使用整數不僅無法表達權重的連續變化，也會讓硬體加速晶片（如 GPU）的矩陣運算變得極為低效。

## 第七部分：比較不同數值表示方式

完成下表。

| 表示方式 | 位元數 | 是否有誤差 | 主要用途 |
|----------|--------|------------|----------|
| int | 無上限 | 無 | 精確計數、索引、邏輯運算與大整數計算 |
| FP64 | 64 | 有 | 高精度科學計算、模擬、工程模擬 |
| FP32 | 32 | 有 | 傳統深度學習模型的訓練與推論標準 |
| FP16 | 16 | 有 | 加速深度學習推論、節省記憶體 |
| BF16 | 16 | 有 | 大型 AI 模型的訓練與推論（兼顧範圍與速度） |

並回答：

1. 哪種表示方式最精確？
    int（針對整數，支援任意精度且完全無誤差）或 FP64（在浮點數中位元數最多、精確度最高）。
2. 哪種表示方式最節省記憶體？
    FP16 與 BF16。兩者皆為 16 位元（僅佔用 2 個位元組），記憶體消耗僅有 FP32 的一半、FP64 的四分之一。
3. 哪種表示方式最常見於 AI 訓練？
    BF16 與 FP32。BF16 是目前大型深度學習與大語言模型訓練的主流（兼顧大數值範圍與高效率）；FP32 則是用於確保傳統模型數值穩定收斂的標準。
4. 哪種表示方式最常見於 AI 推論（Inference）？
    FP16 與 BF16。在模型部署與推論階段，多數會採用 16 位元浮點數來降低記憶體頻寬需求並大幅提升運算速度。

## 第八部分：綜合思考

### 思考題一

Python 官方文件指出：

```python
int
```

具有 Unlimited Precision（無限制精度）。

請回答：

1. Unlimited Precision 是什麼意思？
    這代表 Python 的整數沒有固定的位元數限制（不像傳統 C 或 Java 語言有固定的 32 位元或 64 位元上限），系統會根據數值的大小，動態調整並擴充所需的位數。
2. 它是否代表可以儲存無限大的整數？
    嚴格來說不行。雖然它在邏輯上沒有人為設定的上限，但「物理上」依然受限於電腦可用的實體記憶體（RAM）與虛擬記憶體。只要記憶體夠大就能繼續變大，但若數值大到耗盡記憶體，程式就會引發 MemoryError。
3. CPython 如何實現這項特性？
    CPython 是透過 PyLongObject 物件來實現的。它內部包含一個記錄正負號的欄位，以及一個動態的數位陣列（digit array）。當數值增長超過當前空間時，系統就會自動重新配置（Reallocate）一塊更大的記憶體來存放更多的數位區段。

### 思考題二

比較：

```python
int
```

與

```python
float
```

回答：

1. 為什麼 `int` 幾乎沒有固定上限？
    因為 Python 採用動態記憶體配置（PyLongObject）。當整數數值變大時，系統會自動分配更多的記憶體區段來容納位數，而不是受限於硬體固定的暫存器大小（如 32 或 64 位元）。
2. 為什麼 `float` 卻存在範圍限制？
    因為 float 底層對應 C 語言的 64 位元雙精度浮點數（IEEE 754 標準），其記憶體空間是固定 8 個位元組。因為大小固定、欄位分配明確（固定位元給指數與尾數），所以能表達的數值大小自然存在上限與下限。
3. 為什麼 `float` 可能產生誤差？
    因為電腦底層使用二進位來儲存與運算，而許多常見的十進位小數（例如 0.1）在二進位中會變成無限循環小數。受限於固定的位元長度，電腦必須在某個地方將其「截斷」，因而產生了微小的捨入誤差。

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。
    正確性優先：追求絕對的數學正確，不計記憶體代價。
    效率優先：追求極致的硬體運算速度與省空間，容許微小誤差。
2. 各自的優點與缺點。
    正確性優先（int）：優點是不失真、不溢位；缺點是速度慢、佔用記憶體大。
    效率優先（FP16/BF16）：優點是省空間、運算極快；缺點是會產生捨入誤差。
3. 為什麼大型語言模型通常選擇較低精度的表示方式？
    模型參數動輒數千億，低精度能大幅節省記憶體並加快資料傳輸，且神經網路對微小誤差有高容忍度。
4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？
    記憶體需求暴增 4 倍、硬體頻寬無法負荷、運算速度大幅下降，導致 AI 訓練與推論成本變得極高。

### 挑戰題

請用自己的話解釋：

```python
x = 10**1000
```

為什麼 Python 可以正常運作；

而大型 AI 模型卻通常選擇使用：

```text
FP32
FP16
BF16
```

而不是任意精度整數或高精度浮點數？

請從：

- 記憶體使用量
- 計算速度
- 數值精度
- GPU 運算效率

等角度進行討論。

1. 記憶體使用量 (Memory Usage)
Python int（任意精度）：隨著數字位數增加，記憶體會動態膨脹。像 10**1000 這種大數字需要上百個位元組來儲存。若將這種動態大小的整數套用到擁有數千億參數的 AI 模型中，記憶體消耗將會失控，甚至無法預估。

AI 浮點數格式 (FP32 / FP16 / BF16)：佔用大小是完全固定的（FP32 佔 4 個位元組，FP16/BF16 僅佔 2 個位元組）。這讓數十億甚至數兆個參數的記憶體需求變得可預測且極度緊湊（例如一個 70B 的模型使用 BF16 剛好約需 140GB 記憶體）。


2. 計算速度 (Calculation Speed)
Python int（任意精度）：當位數變得極大時，加減乘除需要透過軟體層面的演算法模擬多位數直式計算（如大數乘法），運算複雜度較高，速度相對緩慢。

AI 浮點數格式 (FP32 / FP16 / BF16)：屬於固定位元運算，現代處理器（CPU/GPU）在硬體層面上可以直接用單一時脈週期完成加法與乘法，運算速度極快。


3. 數值精度 (Numerical Precision)
Python int（任意精度）：追求完全無失真的精確整數運算，絕不允許任何捨入誤差。

AI 浮點數格式 (FP32 / FP16 / BF16)：屬於有限精度（Limited Precision），會伴隨捨入誤差。但神經網路本質上是透過機率、統計與梯度下降來逼近最佳解，對微小的數值雜訊與誤差具有極高的「容錯能力」，因此不需要絕對精確的整數解。


4. GPU 運算效率 (GPU Computation Efficiency)
Python int（任意精度）：其動態記憶體配置與變長結構無法被 GPU 的平行架構有效利用，若強行用 GPU 做大整數運算，效能會極度低落。

AI 浮點數格式 (FP32 / FP16 / BF16)：現代 GPU（如 NVIDIA 的 Tensor Core）是專門為矩陣乘法設計的硬體加速器。它們針對固定位元的浮點數（尤其是 16 位元的 FP16/BF16）進行了極致的電路級優化，能同時平行處理成千上萬個浮點數運算，提供強大的吞吐量。
