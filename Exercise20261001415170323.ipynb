問題：撰寫一Python程式，為使用者輸入年份（可能是西元或民國）找尋其相對應的十二生肖。十二生肖是以十二年為一週期，每一年分別以一種動物來表示猴（monkey）、雞（rooster）、狗(dog)、豬(pig)、鼠(rat)、牛(ox)、虎(tiger)、兔(rabbit)、龍(dragon)、蛇（snake）、馬(horse)，以及羊(sheep) 。

````python
def get_chinese_zodiac():
    # 按照您指定的順序排列的十二生肖清單
    animals = ['猴', '雞', '狗', '豬', '鼠', '牛', '虎', '兔', '龍', '蛇', '馬', '羊']
    
    user_input = input("請輸入年份（例如：2026、民國115、或 115）：").strip()
    
    try:
        # 判斷是否為民國年格式
        if '民國' in user_input or 'ROC' in user_input.upper():
            digits = ''.join(filter(str.isdigit, user_input))
            ad_year = int(digits) + 1911
            era_str = f"民國 {digits} 年"
        else:
            year_val = int(user_input)
            # 若輸入小於 1911 的數字，自動推斷為民國年
            if year_val < 1911:
                ad_year = year_val + 1911
                era_str = f"民國 {year_val} 年"
            else:
                ad_year = year_val
                era_str = f"西元 {year_val} 年"
        
        # 計算生肖索引（2016年為猴年，作為基準 0）
        zodiac_index = ad_year % 12
        zodiac_animal = animals[zodiac_index]
        
        print("\n" + "="*20)
        print(f" 查詢年份：{era_str} (西元 {ad_year} 年)")
        print(f" 對應生肖：{zodiac_animal}")
        print("="*20)
        
    except ValueError:
        print("❌ 輸入格式錯誤，請輸入純數字或包含「民國」的年份！")

if __name__ == "__main__":
    get_chinese_zodiac()
````
