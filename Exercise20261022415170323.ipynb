## 二、題目說明（Problem Description）

請設計一個 Python 程式，建立學生分數分析系統。

程式必須能夠接收多筆分數，根據指定的倍率（multiplier）及加分值（bonus）調整分數，接著計算調整後的分數平方、總分及平均分數，最後回傳分析結果。

### Part 1：完成函數

請閱讀以下函數，理解各部分的功能。

```python
def analyze_scores(*scores, **options):
    """
    分析學生分數。

    Parameters:
        *scores: 接收不定數量的分數
        **options: 接收選擇性設定，例如 multiplier 和 bonus

    Returns:
        一個字典，包含調整後的分數、平方、總分及平均分數。
    """
    multiplier = options.get("multiplier", 1)
    bonus = options.get("bonus", 0)

    adjusted = [
        score * multiplier + bonus
        for score in scores
    ]

    squared = [
        score ** 2
        for score in adjusted
    ]

    total = sum(adjusted)
    average = total / len(adjusted) if adjusted else 0

    return {
        "adjusted_scores": adjusted,
        "squared_scores": squared,
        "total": total,
        "average": average
    }
```

### Part 2：呼叫函數

已知下列資料：

```python
scores = [60, 70, 80]

settings = {
    "multiplier": 1.1,
    "bonus": 5
}
```

**問題 1：** 請使用 `*` 與 `**` 解包 `scores` 和 `settings`，呼叫 `analyze_scores()` 函數，並將結果存入變數 `result`。

```python
result = analyze_scores(*scores, **settings)
```

**問題 2：** 請印出 `result`，觀察函數回傳的結果。

```python
print(result)
# 預期輸出：
# {'adjusted_scores': [71.0, 82.0, 93.0], 'squared_scores': [5041.0, 6724.0, 8649.0], 'total': 246.0, 'average': 82.0}
```

### Part 3：思考與分析（Questions）

1. 在函數內部，`scores` 的資料型態是什麼？其值為何？
   
   在函數內部，`scores` 的資料型態是 **`tuple`** (元組)；在此範例呼叫中，其值為 `(60, 70, 80)`。

2. 在函數內部，`options` 的資料型態是什麼？其值為何？

3. `score * multiplier` 與 `score ** 2` 有什麼不同？請說明兩個運算子的意義。
   
   - `*` 是 **乘法運算子**：代表進行相乘，在此表示將原始分數乘以指定的放大倍率。
   - `**` 是 **次方 (冪) 運算子**：代表將一個數進行自乘（指數運算），在此表示計算調整後分數的平方。

4. 為什麼函數使用 `return`，而不是只使用 `print()`？
   
   因為 `print()` 只能把計算結果單純顯示在螢幕上，無法讓程式碼的其他部分重複或後續利用；而 `return` 會將結果傳回給呼叫端，讓我們能將回傳的字典賦值給變數 `result` 進一步作其他運算或儲存。

7. 如果沒有傳入任何分數，程式會回傳什麼結果？請解釋程式如何處理這種情況。

## 三、語法整理（Learning Focus）

| 語法 | 用途 |
|---|---|
| `*scores`（函數定義） | 將不定數量的位置參數收集成 tuple |
| `**options`（函數定義） | 將不定數量的關鍵字參數收集成 dictionary |
| `score * multiplier` | 乘法 |
| `score ** 2` | 次方 |
| `f(*scores)` | 將可迭代物件解包為位置參數 |
| `f(**settings)` | 將字典解包為關鍵字參數 |
| `return` | 將函數的結果回傳給呼叫端 |

### 總結回覆：

在 Python 中，`*`、`**` 與 `return` 扮演著核心且多用途的角色。在不同的位置（例如定義函數時 vs. 呼叫函數時，或者一般的算術運算中），它們具有截然不同的含意：

1. **`*`（單星號）**
   - **定義函數時（`*args`）**：作為「參數收集器」，會將多個位置參數打包成一個 `tuple`。例如：`def f(*args):`。
   - **呼叫函數時（解包）**：作為「解包運算子」，將可迭代物件（如串列、元組）拆解為獨立的單一位置參數。例如：`f(*[1, 2])` 等同於 `f(1, 2)`。
   - **數值運算時**：代表「乘法運算子」。例如：`3 * 5` 等於 `15`。

2. **`**`（雙星號）**
   - **定義函數時（`**kwargs`）**：作為「參數收集器」，會將多個命名關鍵字參數打包成一個 `dict`。例如：`def f(**kwargs):`。
   - **呼叫函數時（解包）**：將字典拆解為獨立的關鍵字對應傳入函數。例如：`f(**{'a': 1})` 等同於 `f(a=1)`。
   - **數值運算時**：代表「次方（冪）運算子」。例如：`3 ** 2` 等於 `9`。

3. **`return`（回傳值）**
   - 結束函數並將一個或多個結果值傳遞回呼叫它的地方，讓其結果可以被存入變數或作後續處理（與單純使用 `print()` 僅輸出至螢幕不同）。

#### 程式碼範例展示：
```python
# 範例函數定義，同時展現收集器與 return 結構
def demo_func(*args, **kwargs):
    # 1. 此時的 *args 是收集器，**kwargs 也是收集器
    # 2. 我們在此處做數值計算並使用 return 將結果傳出
    return args, kwargs

# 範例呼叫，展現解包運算子
my_list = [1, 2]
my_dict = {"name": "Alice"}

# 此時的 * 與 ** 是解包功能
result = demo_func(*my_list, **my_dict)
print(result) # 輸出: ((1, 2), {'name': 'Alice'})
```
